```python
import pandas as pd
from urllib.parse import urlparse
import json
```

```python
df = pd.read_csv('khmer_news_COMPLETE_20251112_031508.csv')
```

```python
OUT = "dab_news.json"
records = []
for _, row in df.iterrows():
    url = row.get("Source_URL", "") or row.get("source", "")
    title = row.get("Title", "") or row.get("title", "")
    content = row.get("Content", "") or row.get("content", "")

    # build source from URL
    source = ""
    if url:
        p = urlparse(url)
        source = f"{p.scheme}://{p.netloc}" if p.scheme else (p.netloc or url)

    # parse dates and format
    pub_raw = row.get("Date", "") or row.get("publication_date", "")
    scr_raw = row.get("Scraped_At", "") or row.get("scrape_date", "")

    pub_dt = pd.to_datetime(pub_raw, errors="coerce")
    pub_str = "" if pd.isna(pub_dt) else pub_dt.strftime("%m/%d/%Y")

    scr_dt = pd.to_datetime(scr_raw, errors="coerce")
    scr_str = "" if pd.isna(scr_dt) else scr_dt.strftime("%m/%d/%Y %H:%M")

    records.append({
        "url": url,
        "title": title,
        "content": content,
        "source": source,
        "publication_date": pub_str,
        "scrape_date": scr_str
    })

with open(OUT, "w", encoding="utf-8") as f:
    json.dump(records, f, ensure_ascii=False, indent=2)

print(f"Saved {len(records)} items to {OUT}")

from urllib.parse import urlparse
import json

CSV = "khmer_news_COMPLETE_20251112_031508.csv"
OUT = "khmer_news_formatted.json"

# read CSV (keep as strings, avoid NA -> empty string)
df = pd.read_csv(CSV, dtype=str, keep_default_na=False)

records = []
for _, row in df.iterrows():
    url = row.get("Source_URL", "") or row.get("source", "")
    title = row.get("Title", "") or row.get("title", "")
    content = row.get("Content", "") or row.get("content", "")

    # build source from URL
    source = ""
    if url:
        p = urlparse(url)
        source = f"{p.scheme}://{p.netloc}" if p.scheme else (p.netloc or url)

    # parse dates and format
    pub_raw = row.get("Date", "") or row.get("publication_date", "")
    scr_raw = row.get("Scraped_At", "") or row.get("scrape_date", "")

    pub_dt = pd.to_datetime(pub_raw, errors="coerce")
    pub_str = "" if pd.isna(pub_dt) else pub_dt.strftime("%m/%d/%Y")

    scr_dt = pd.to_datetime(scr_raw, errors="coerce")
    scr_str = "" if pd.isna(scr_dt) else scr_dt.strftime("%m/%d/%Y %H:%M")

    records.append({
        "url": url,
        "title": title,
        "content": content,
        "source": source,
        "publication_date": pub_str,
        "scrape_date": scr_str
    })

with open(OUT, "w", encoding="utf-8") as f:
    json.dump(records, f, ensure_ascii=False, indent=2)

print(f"Saved {len(records)} items to {OUT}")
```
